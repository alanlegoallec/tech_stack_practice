```python
# --- 1. Imports ---
import boto3
import pandas as pd
import io
import matplotlib.pyplot as plt
```

```python
# --- 2. S3 Configuration ---
BUCKET_NAME = "fullstack-practice-data"
OBJECT_KEY = "outputs/multiplication_log.csv"
```

```python
# --- 3. Load from S3 ---
session = boto3.Session(profile_name="deployer-full-stack-practice")
s3 = session.client("s3")

response = s3.get_object(Bucket=BUCKET_NAME, Key=OBJECT_KEY)
df = pd.read_csv(io.BytesIO(response["Body"].read()))

print("✅ Loaded DataFrame:")
display(df.head())
```

```python
# --- 4. Data Summary ---
print("📊 Summary Stats:")
display(df.describe(include="all"))
```

```python
# --- 5. Plot: Input vs. Product ---
plt.figure(figsize=(8, 5))
plt.scatter(df["user_input"], df["product"], alpha=0.7)
plt.xlabel("User Input")
plt.ylabel("Product")
plt.title("User Input vs. Product")
plt.grid(True)
plt.show()
```

```python
# --- 6. Latency Distribution ---
plt.figure(figsize=(8, 5))
df["inference_latency_ms"].hist(bins=20)
plt.xlabel("Latency (ms)")
plt.ylabel("Frequency")
plt.title("Latency Distribution")
plt.grid(True)
plt.show()
```

```python
# --- 7. Most Frequent Names ---
print("👤 Most common names:")
display(df["associated_name"].value_counts().head())
```
